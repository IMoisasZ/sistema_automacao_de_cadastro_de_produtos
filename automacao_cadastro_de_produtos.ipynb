# Automação

## Sistema de automação de cadastro de produtos

## Bibliotecas

```python
import time
import pyautogui
import pandas as pd
```

## Variaveis global

```python
url = "https://dlp.hashtagtreinamentos.com/python/intensivao/login"
email = "moises@email.com"
senha = "123456"
```

## Tempo de execução de cada etapa

```python
pyautogui.PAUSE = 2
```

## Iniciando as etapas

### 1.0 Abrir o sistema

```python
# Preciona a tecla windows
pyautogui.press("win")
# Escreve o nome do navegador que vai utilizar
pyautogui.write("chrome")
# Preciona a tecla enter para abrir o navegador
pyautogui.press("enter")
# Clica na barra de endereços
pyautogui.click(x=430, y=72)
# Escreve o endereço do sistema
pyautogui.write(url)
# Preciona enter para entrar no sistema
pyautogui.press("enter")
# Espera um tempo até que a pagina seja carregada
time.sleep(3)
# Move o cursor para o campo de email
pyautogui.click(x=737, y=515)
```

### 2.0 Fazendo login

```python
# Informa o email
pyautogui.write(email)
# Vai para o campo de senha
pyautogui.press("tab")
# Informa a senha
pyautogui.write(senha)
# Vai para o botão de enviar
pyautogui.press("tab")
# Preciona enter para fazer o login
pyautogui.press('enter')
# Esperar um tempo para entrar no sistema
time.sleep(3)
```

### 3.0 Abrindo a base de dados

```python
tabela = pd.read_csv("produtos.csv")
```

### 4.0 Cadastrando produtos

```python
# Loop para percorrer todos as linhas da tabela
for linha in tabela.index:
    # Move o cursor para o campo codigo do produto
    pyautogui.click(x=666, y=367)
    # Informar o codigo do produto
    codigo_produto = str(tabela.loc[linha, "codigo"])
    pyautogui.write(codigo_produto)
    # Ir para o próximo campo
    pyautogui.press("tab")
    # Informar a marca do produto
    marca = str(tabela.loc[linha, 'marca'])
    pyautogui.write(marca)
    # Ir para o proximo campos
    pyautogui.press("tab")
    # Informar o tipo
    tipo = str(tabela.loc[linha, 'tipo'])
    pyautogui.write(tipo)
    # Ir para o proximo campo
    pyautogui.press("tab")
    # Informar a categoria
    categoria = str(tabela.loc[linha, 'categoria'])
    pyautogui.write(categoria)
    # Ir para o proximo campo
    pyautogui.press("tab")
    # Informaro campo preço unitário
    preco_unitario = str(tabela.loc[linha, 'preco_unitario'])
    pyautogui.write(preco_unitario)
    # Ir para o proximo campo
    pyautogui.press("tab")
    # Informar o custo
    custo = str(tabela.loc[linha, 'custo'])
    pyautogui.write(custo)
    # Ir para o proximo campo
    pyautogui.press("tab")
    # Informar a Obs
    obs = str(tabela.loc[linha, 'obs'])
    if obs != "nan":
        pyautogui.write(obs)
    # Ir para o botão de incluir
    pyautogui.press("tab")
    # Precionar o botão para incluir
    pyautogui.press("enter")
    # Esperar um tempo para inclusão acontecer
    time.sleep(1    )
    # Mover o curso par ao incio da tela
    pyautogui.scroll(5000)
```
